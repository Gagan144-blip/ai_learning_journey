```python
data = {
    'a': np.array([2, 3, 4]),
    'b': np.array([4, 5, 4]),
    'c': np.array([3,4 ,5])
}
df = pd.DataFrame(data)
print(df)
```

```python
# setting a custom index 
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}

df = pd.DataFrame(data)
res = df.set_index('Age')
print(res)
```

```python
# to reset index to default
res = df.reset_index(drop=True)
print(res)
```

```python
# indexing with loc
df = pd.DataFrame(data, index=['A', 'B', 'C'])
print(df)
print(df.loc['A'])
```

```python
# accessing row by index
# iloc -> for positional indexing
# loc -> for label based indexing
second_row = df.iloc[0]
print(second_row)
```

```python
# access multiple rows
multiple_rows = df.loc['A':'C', ['Name', 'Age']]
print(multiple_rows)
```

```python
# access value at specific cell using at and iat
age_at_index = df.at['A', 'Age']
print(age_at_index)
```

```python
# iat
Name_at_index = df.iat[2, 1]
print(Name_at_index)
```
