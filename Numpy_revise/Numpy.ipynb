```python
# Data types in numpy
x = np.array([1, 2])
print(x.dtype)

y = np.array([0.4, 3.8])
print(y.dtype)

x = np.array([4, 5], dtype= np.int64)
print(x.dtype)
```

```python
# maths operations on datatype array

arr1  = np.array([[4, 5], [6, 7], [7, 8]], dtype = np.float64)
arr1
```

```python
# fill array with zeros

arr = np.zeros((3, 4))
print(arr)
```

```python
# .randint used to create a random array in a specific range of numbers

arr_int = np.random.randint(1 , 10, size=(4, 5))
arr_int
```
